## 💻 Enhanced Data Transformation Example

Building on the original chapter's data transformation example, here's an improved version with better structure and error handling:

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

import pandas as pd
import numpy as np
from datetime import datetime, timedelta
from typing import Optional
import warnings
warnings.filterwarnings('ignore')

class DataTransformer:
    """
    Enhanced data transformation class with error handling and validation.
    """
    
    def __init__(self, random_seed: int = 42):
        np.random.seed(random_seed)
    
    def generate_random_date(self, start_date: datetime, end_date: datetime) -> str:
        """
        Generate a random date between start and end dates.
        """
        if start_date >= end_date:
            raise ValueError("Start date must be before end date")
        
        days_diff = (end_date - start_date).days
        random_days = np.random.randint(0, days_diff + 1)
        random_date = start_date + timedelta(days=random_days)
        
        return random_date.strftime('%Y-%m-%d')
    
    def generate_diverse_scores(self, n_samples: int) -> np.ndarray:
        """
        Generate diverse propensity scores with realistic distribution.
        """
        scores = []
        
        for _ in range(n_samples):
            p = np.random.rand()
            
            if p < 0.4:  # 40% extreme values
                score = np.random.beta(0.5, 0.5)
            elif p < 0.7:  # 30% moderate values
                score = np.random.normal(0.5, 0.15)
            else:  # 30% uniform distribution
                score = np.random.rand()
            
            scores.append(score)
        
        return np.clip(scores, 0, 1)
    
    def create_sample_dataset(self, n_rows: int = 10000) -> pd.DataFrame:
        """
        Create a realistic sample dataset.
        """
        data = {
            'customer_id': range(1, n_rows + 1),
            'age': np.random.randint(18, 80, n_rows),
            'income': np.random.normal(50000, 15000, n_rows),
            'region': np.random.choice(['North', 'South', 'East', 'West'], n_rows),
            'product_interest': np.random.choice(['A', 'B', 'C', 'D'], n_rows)
        }
        
        df = pd.DataFrame(data)
        df['income'] = np.clip(df['income'], 20000, 150000)
        
        return df
    
    def transform_dataset(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Transform dataset with enhanced propensity scores and dates.
        """
        if df.empty:
            raise ValueError("Input DataFrame is empty")
        
        result_df = df.copy()
        
        # Generate random dates
        start_date = datetime(2024, 1, 1)
        end_date = datetime(2024, 12, 31)
        
        result_df['propensity_date'] = [
            self.generate_random_date(start_date, end_date) 
            for _ in range(len(df))
        ]
        
        # Generate diverse scores
        result_df['propensity_score'] = self.generate_diverse_scores(len(df))
        
        # Add score categories
        result_df['score_category'] = pd.cut(
            result_df['propensity_score'], 
            bins=[0, 0.3, 0.7, 1.0], 
            labels=['Low', 'Medium', 'High']
        )
        
        return result_df
    
    def generate_summary(self, df: pd.DataFrame) -> None:
        """
        Generate transformation summary.
        """
        print("\n📊 TRANSFORMATION SUMMARY")
        print("=" * 50)
        print(f"Total records: {len(df):,}")
        
        if 'propensity_score' in df.columns:
            score_stats = df['propensity_score'].describe()
            print(f"\nScore Statistics:")
            print(f"  Mean: {score_stats['mean']:.4f}")
            print(f"  Std:  {score_stats['std']:.4f}")
            print(f"  Min:  {score_stats['min']:.4f}")
            print(f"  Max:  {score_stats['max']:.4f}")
        
        if 'score_category' in df.columns:
            dist_counts = df['score_category'].value_counts()
            print(f"\nScore Distribution:")
            for category, count in dist_counts.items():
                percentage = (count / len(df)) * 100
                print(f"  {category}: {count:,} ({percentage:.1f}%)")

# Demonstration
print("🚀 ENHANCED DATA TRANSFORMATION DEMONSTRATION")

transformer = DataTransformer(random_seed=42)

try:
    # Create and transform data
    sample_df = transformer.create_sample_dataset(10000)
    print(f"\nCreated sample dataset with {len(sample_df):,} rows")
    
    result_df = transformer.transform_dataset(sample_df)
    print("✅ Data transformation completed!")
    
    # Show results
    transformer.generate_summary(result_df)
    
    print("\n📋 Sample of Transformed Data:")
    columns_to_show = ['customer_id', 'age', 'region', 'propensity_score', 'score_category']
    print(result_df[columns_to_show].head())
    
except Exception as e:
    print(f"❌ Error: {e}")
```

## 🔍 IPython and Jupyter Features

Essential features for efficient data analysis:

```python
import time
import sys
import matplotlib.pyplot as plt

print("🔍 IPYTHON FEATURES GUIDE\n")

print("1. Object Inspection:")
print("   pd.DataFrame? - Show docstring")
print("   pd.DataFrame?? - Show source code")
print("   Use Tab completion for method discovery")

print("\n2. Magic Commands:")
print("   %time - Time single execution")
print("   %timeit - Time multiple executions")
print("   %debug - Enter debugger")
print("   %who - List variables")

print("\n3. Performance Comparison Example:")

# Pure Python vs NumPy
def python_sum(data):
    return sum(x**2 for x in data)

def numpy_sum(data):
    return np.sum(data**2)

test_data = list(range(100000))
np_data = np.array(test_data)

# Time both approaches
start = time.time()
python_result = python_sum(test_data)
python_time = time.time() - start

start = time.time()
numpy_result = numpy_sum(np_data)
numpy_time = time.time() - start

print(f"   Pure Python: {python_time:.6f}s")
print(f"   NumPy: {numpy_time:.6f}s")
print(f"   NumPy is {python_time/numpy_time:.1f}x faster")

print("\n4. Matplotlib Integration:")

# Create sample visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)

ax1.plot(x, y, 'b-', alpha=0.7)
ax1.set_title('Sine Wave')
ax1.grid(True, alpha=0.3)

ax2.hist(np.random.randn(1000), bins=30, alpha=0.7, color='green')
ax2.set_title('Random Distribution')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("   ✅ Inline plots generated")
```

## 🐍 Python Language Fundamentals

Core concepts essential for data analysis:

```python
from collections import namedtuple

print("🐍 PYTHON FUNDAMENTALS\n")

print("1. Dynamic Typing and References:")

# Variable references
a = [1, 2, 3]
b = a  # Same object
c = a.copy()  # New object

print(f"   Original: {a}")
print(f"   Reference: {b}")
print(f"   Copy: {c}")
print(f"   a is b: {a is b}")
print(f"   a is c: {a is c}")

a.append(4)
print(f"   After append - a: {a}, b: {b}, c: {c}")

print("\n2. Data Structures for Analysis:")

# Lists for sequences
data_points = [23, 45, 67, 89, 12]
squared = [x**2 for x in data_points]
filtered = [x for x in data_points if x > 50]

print(f"   Original: {data_points}")
print(f"   Squared: {squared}")
print(f"   Filtered: {filtered}")

# Dictionaries for key-value data
sales_data = {
    'Q1': [100, 120, 110],
    'Q2': [130, 140, 125],
    'Q3': [150, 160, 145]
}

quarterly_totals = {quarter: sum(values) for quarter, values in sales_data.items()}
print(f"   Quarterly totals: {quarterly_totals}")

# Sets for unique collections
categories_a = {'electronics', 'books', 'clothing'}
categories_b = {'books', 'sports', 'electronics'}

print(f"   Union: {categories_a | categories_b}")
print(f"   Intersection: {categories_a & categories_b}")

# Tuples for structured data
employee = ('Alice', 30, 'Manager', 85000)
name, age, position, salary = employee
print(f"   Employee: {name}, {age} years, {position}, ${salary:,}")

# Named tuples
Employee = namedtuple('Employee', ['name', 'age', 'position', 'salary'])
emp = Employee('Bob', 28, 'Developer', 70000)
print(f"   Named tuple: {emp.name}, {emp.position}")

print("\n3. Memory Management:")
data = [1, 2, 3, 4, 5]
print(f"   Reference count: {sys.getrefcount(data)}")

data_ref = data
print(f"   After creating reference: {sys.getrefcount(data)}")

del data_ref
print(f"   After deleting reference: {sys.getrefcount(data)}")
```
