## 🔹 Tuplas

O autor começa explorando **tuplas**, uma estrutura sequencial que, diferente das listas, é **imutável** (uma vez criada, seus elementos não podem ser alterados).  

Apesar disso, tuplas podem conter objetos mutáveis dentro delas (como listas), permitindo alterações indiretas. Outro ponto importante é que qualquer sequência ou iterador pode ser convertido em tupla através da função `tuple()`.  

### 💡 Principais conceitos sobre tuplas:
- Criadas normalmente com parênteses `()`  
- Permitem **concatenação** e **operações numéricas**  
- Suportam **indexação** e **fatiamento** (assim como listas)  
- Podem utilizar `*rest` para capturar múltiplos valores em desempacotamento  
- Possuem apenas dois métodos principais:  
  - `count()` → Conta quantas vezes um valor aparece  
  - `index()` → Retorna a posição do primeiro valor encontrado  

### ⚠️ Limitações:
Como são imutáveis, tuplas **não possuem métodos de modificação** como `append()`, `remove()` ou `sort()`.  

--- 


```python
ind = tuple([4,8,2, 16, 24])
print(ind)
```

Há a possibilidade de ter um objeto mutável dentro de uma tupla. Por exemplo, uma lista:
Sendo necessário apenas indicar a indexação correta do elemento mutável. (Indexação em Python iinicia sempre em 0(zero))

```python
ex = ("foo", [4,5,6,8], "end")
ex[1].append(99)
print(ex)
```

É possivel tambem utilizar concatenaçções ou expressões numéricas entre tuplas

```python
#Concatenações 
a = (5, 10, 15)
b = (1, 2, 3)
c = a + b
print(f'Concatenação',{c})

#Expressões numéricas
d = a * 3
print(f'Expressão Numérica:',{d})

#Indexação e fatiamento
print(f'Indexação:',{c[0]})
print(f'Fatiamento:',{c[0:3]})
```

Para extrair valores dentro de uma tupla pode-se usar *rest
Ao direcionar os valores necessários para extrair da tupla o uso de rest vai transformar o restante numa lista.

```python
a = (1, 2, 3, 5, 9)
x, y, *rest = a
print(x, y, rest)
```

Tuplas são estruturas imutáveis, o que limita alguns métodos disponíveis em listas, por exemplo. Além de count(), tuplas possuem apenas dois métodos principais:

count(): Conta quantas vezes um valor aparece.
index(): Encontra a primeira posição de um valor na tupla.
Como tuplas são imutáveis, não é possível modificar seus elementos diretamente, ou seja, não há métodos como append(), remove(), sort(), etc.
